```python
import requests

def search_openalex(keywords: str, max_results: int = 5):
    """
    Search OpenAlex API for papers matching the keywords.
    Args:
        keywords: Search query (e.g., "climate change mitigation").
        max_results: Maximum number of results to return.
    Returns:
        List of dictionaries with paper metadata.
    """
    base_url = "https://api.openalex.org/works"
    params = {
        "search": keywords,
        "per-page": max_results,
    }
    response = requests.get(base_url, params=params)
    response.raise_for_status()
    data = response.json()
    return data.get("results", [])

def print_results(results):
    """Print basic metadata for each result."""
    for idx, paper in enumerate(results, 1):
        print(f"{idx}. {paper.get('title')}")
        print(f"   DOI. {paper.get('doi')}")
        print(f"   Year: {paper.get('publication_year')}")
        print(f"   OpenAlex ID: {paper.get('id')}")
        print(f"   Link: {paper.get('primary_location', {}).get('landing_page_url')}\n")

# Example usage
keywords = """
("risk assessment" OR "risk analysis") AND
("Offshore windfarm" OR "Offshore windmill" OR "Offshore wind turbine" OR "Offshore wind park" OR "OWF") AND
("maritime" OR  "navigational" OR "nautical")
"""
results = search_openalex(keywords)
print(f"Found {len(results)} papers for '{keywords}':\n")
print_results(results)
```

```python
from pyalex import Works
query_openAlex = """
("risk assessment") AND ('Static') AND ('UK') AND
("Offshore windfarm") AND
("maritime" OR  "navigational" OR "nautical")
"""
def get_paper_count(keywords: str) -> Works:
    """
    Get the total number of papers matching the keywords using pyalex.
    Args:
        keywords: Search query (e.g., "climate change mitigation").
    Returns:
        Total number of matching papers.
    """
    works = Works().search(keywords)
    return works

# Example usage

work = get_paper_count(query_openAlex)
print(f"Total papers found for '{query_openAlex}': {work.count()}")
```
