```python
# Run this app with `python dash_app.py` and
# visit http://127.0.0.1:8050/ in your web browser.

import pandas as pd
from dash import Dash, dash_table, html

app = Dash(__name__)

# Create fake dataframe with the 9 required columns
df = pd.DataFrame(
    {
        "A": [1, 0, 1, 0, 1],
        "B": [0, 1, 0, 1, 0],
        "C": [1, 1, 0, 0, 1],
        "A_explain": [
            "Explanation for A being 1",
            "Explanation for A being 0",
            "Another explanation for A being 1",
            "Another explanation for A being 0",
            "Final explanation for A being 1",
        ],
        "B_explain": [
            "Explanation for B being 0",
            "Explanation for B being 1",
            "Another explanation for B being 0",
            "Another explanation for B being 1",
            "Final explanation for B being 0",
        ],
        "C_explain": [
            "Explanation for C being 1",
            "Explanation for C being 1",
            "Another explanation for C being 0",
            "Another explanation for C being 0",
            "Final explanation for C being 1",
        ],
        "user": [1, 0, 1, 0, 1],
        "title": [
            "Title of the first paper",
            "Title of the second paper",
            "Title of the third paper",
            "Title of the fourth paper",
            "Title of the fifth paper",
        ],
        "consensus": [1.0, 0.8, 0.3, 0.1, 0.95],
    }
)

# Sort by consensus score (descending - highest first)
df = df.sort_values("consensus", ascending=False).reset_index(drop=True)

# Binary columns to color (green for 1, red for 0)
binary_columns = ["A", "B", "C", "user"]

# Build conditional styling
style_data_conditional = []

# Binary columns: green for 1, red for 0
for col in binary_columns:
    style_data_conditional.append(
        {
            "if": {"filter_query": f"{{{col}}} = 1", "column_id": col},
            "backgroundColor": "#90EE90",
            "color": "black",
        }
    )
    style_data_conditional.append(
        {
            "if": {"filter_query": f"{{{col}}} = 0", "column_id": col},
            "backgroundColor": "#FFCCCB",
            "color": "black",
        }
    )

# Consensus column: gradient from red (0) to green (1)
# Using multiple conditions to simulate a gradient
style_data_conditional.append(
    {
        "if": {"filter_query": "{consensus} < 0.2", "column_id": "consensus"},
        "backgroundColor": "#FF0000",
        "color": "white",
    }
)
style_data_conditional.append(
    {
        "if": {
            "filter_query": "{consensus} >= 0.2 && {consensus} < 0.4",
            "column_id": "consensus",
        },
        "backgroundColor": "#FF4500",
        "color": "white",
    }
)
style_data_conditional.append(
    {
        "if": {
            "filter_query": "{consensus} >= 0.4 && {consensus} < 0.6",
            "column_id": "consensus",
        },
        "backgroundColor": "#FFA500",
        "color": "black",
    }
)
style_data_conditional.append(
    {
        "if": {
            "filter_query": "{consensus} >= 0.6 && {consensus} < 0.8",
            "column_id": "consensus",
        },
        "backgroundColor": "#9ACD32",
        "color": "black",
    }
)
style_data_conditional.append(
    {
        "if": {"filter_query": "{consensus} >= 0.8", "column_id": "consensus"},
        "backgroundColor": "#00FF00",
        "color": "black",
    }
)

app.layout = html.Div(
    children=[
        html.H1(children="SLR-RAG Data Table"),
        html.Div(
            children="""
        Dash table displaying the review data.
    """
        ),
        dash_table.DataTable(
            id="data-table",
            columns=[{"name": col, "id": col} for col in df.columns],
            data=df.to_dict("records"),
            style_table={"overflowX": "auto"},
            style_cell={
                "textAlign": "left",
                "padding": "10px",
                "whiteSpace": "normal",
                "height": "auto",
            },
            style_header={
                "backgroundColor": "rgb(230, 230, 230)",
                "fontWeight": "bold",
            },
            style_data_conditional=style_data_conditional,
            page_size=10,
        ),
    ]
)

app.run(jupyter_mode="external")
```
