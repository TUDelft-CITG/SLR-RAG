```python
import pandas as pd

from slrrag.settings import OUTPUT_CSV_TO_MANUAL_FILTER

df_analyisis = pd.read_csv(OUTPUT_CSV_TO_MANUAL_FILTER).head(5)
df_analyisis
```

```python
def test_map_func(input: pd.Series):
    get_d = {"t": input["title"].lower(), "d": pd.isna(input["doi"])}
    return pd.Series(get_d)


pd.concat((df_analyisis, df_analyisis.apply(test_map_func, axis=1)), axis=1)
```
