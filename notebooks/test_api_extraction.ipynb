```python
import ast


def explode_and_groupby_topic(
    df_input: pd.DataFrame, name_topic_col: str, min_count_to_keep=100
):
    df_input = (
        df_input[df_input[name_topic_col].notna()]
        .reset_index()[[name_topic_col, "is_reference"]]
        .copy()
    )
    df_input[name_topic_col] = df_input[name_topic_col].apply(ast.literal_eval)
    df_input = df_input.explode(name_topic_col)

    df_common_topic_common = (
        df_input.groupby([name_topic_col])
        .size()
        .sort_values(ascending=False)
        .rename("count")
        .reset_index()
    )
    df_common_topic_common = df_common_topic_common[
        (df_common_topic_common["count"] > min_count_to_keep)
    ]

    df_common_topic_splitted = (
        df_input.groupby([name_topic_col, "is_reference"])
        .size()
        .sort_values(ascending=False)
        .rename("count")
        .reset_index()
    )
    df_common_topic_splitted = df_common_topic_splitted[
        df_common_topic_splitted[name_topic_col].isin(
            df_common_topic_common[name_topic_col].unique()
        )
    ]
    return df_common_topic_splitted


px.bar(
    explode_and_groupby_topic(
        df_recent, name_topic_col="topic_lens", min_count_to_keep=100
    ),
    x="topic_lens",
    color="is_reference",
    y="count",
)
```

```python
import pandas as pd
import pandera.pandas as pa
from pandera.typing.pandas import DataFrame, Series


class InputSchema(pa.DataFrameModel):
    year: Series[int] = pa.Field(gt=2000, coerce=True)
    month: Series[int] = pa.Field(ge=1, le=12, coerce=True)


class OutputSchema(InputSchema):
    revenue: Series[float]


@pa.check_types(lazy=PANDERA_LAZY_VALIDATION)
def transform(df: DataFrame[InputSchema]) -> DataFrame[OutputSchema]:
    return df.assign(revenue=100.0)


df_papers = pd.DataFrame(
    {
        "year": ["2001", "2002", "2003"],
        "month": ["3", "6", "9"],
        "day": ["200", "156", "365"],
    }
)

return_df: DataFrame[OutputSchema] = transform(df_papers)

invalid_df = pd.DataFrame(
    {
        "year": ["2001", "2002", "1999"],
        "month": ["3", "6", "12"],
        "day": ["200", "156", "365"],
    }
)

try:
    transform(invalid_df)
except pa.errors.SchemaError as exc:
    print(exc)
```

```python
import pandera.pandas as pa
from datetime import datetime
from typing import Optional
from pandera.typing import Series, Int64

# New global variables for missing columns
NAME_COL_IS_FOUND = "is_found"
NAME_COL_YEAR = "year"
NAME_COL_ID_SCOPUS = "id_scopus"
NAME_COL_ID_OPENalex = "id_openalex"
NAME_COL_ID_LENS = "id_lens"
NAME_COL_ID_DIMENSIONS = "id_dimensions"
NAME_COL_IN_SCOPUS = "in_scopus"
NAME_COL_IN_OPENalex = "in_openalex"
NAME_COL_IN_LENS = "in_lens"
NAME_COL_IN_DIMENSIONS = "in_dimensions"
NAME_COL_DOI = "doi"
NAME_COL_TITLE = "title"
NAME_COL_DATE = "date"
NAME_COL_ABSTRACT = "abstract"
NAME_COL_TYPE = "type"
NAME_COL_PDF = "pdf"
NAME_COL_IS_REFERENCE = "is_reference"


class SchemaPapers(pa.DataFrameModel):
    hello: Series[str] = pa.Field(nullable=True, alias=NAME_COL_DOI)
    title: Series[str] = pa.Field(nullable=True, alias=NAME_COL_TITLE)
    date: Series[datetime] = pa.Field(nullable=True, alias=NAME_COL_DATE)
    abstract: Series[str] = pa.Field(nullable=True, alias=NAME_COL_ABSTRACT)
    is_reference: Series[bool] = pa.Field(nullable=True, alias=NAME_COL_IS_REFERENCE)
    is_found: Series[bool] = pa.Field(nullable=True, alias=NAME_COL_IS_FOUND)
    year: Series[Int64] = pa.Field(nullable=True, alias=NAME_COL_YEAR)

    # Optional fields
    id_scopus: Optional[Series[str]] = pa.Field(nullable=True, alias=NAME_COL_ID_SCOPUS)
    id_openalex: Optional[Series[str]] = pa.Field(
        nullable=True, alias=NAME_COL_ID_OPENalex
    )
    id_lens: Optional[Series[str]] = pa.Field(nullable=True, alias=NAME_COL_ID_LENS)
    id_dimensions: Optional[Series[str]] = pa.Field(
        nullable=True, alias=NAME_COL_ID_DIMENSIONS
    )
    in_scopus: Optional[Series[bool]] = pa.Field(
        nullable=True, alias=NAME_COL_IN_SCOPUS
    )
    in_openalex: Optional[Series[bool]] = pa.Field(
        nullable=True, alias=NAME_COL_IN_OPENalex
    )
    in_lens: Optional[Series[bool]] = pa.Field(nullable=True, alias=NAME_COL_IN_LENS)
    in_dimensions: Optional[Series[bool]] = pa.Field(
        nullable=True, alias=NAME_COL_IN_DIMENSIONS
    )


SchemaPapers.validate(df_papers,lazy=PANDERA_LAZY_VALIDATION).dtypes
```

```python
# Type categories
from pandera.typing import Categorical


TYPE_CATEGORIES = ["article", "book", "report", "other", "conference paper"]

# Mapping from possible input values to canonical categories
TYPE_MAPPING = {
    # Article variations
    "article": "article",
    "journal article": "article",
    "journal": "article",
    "paper": "article",
    "journal paper": "article",
    # Book variations
    "book": "book",
    "book chapter": "book",
    "monograph": "book",
    "chapter": "book",
    # Report variations
    "report": "report",
    "technical report": "report",
    "working paper": "report",
    "preprint": "report",
    # Conference variations
    "conference paper": "conference paper",
    "conference": "conference paper",
    "proceedings paper": "conference paper",
    "proceedings": "conference paper",
    # Other
    "other": "other",
}


class SchemaPapers(pa.DataFrameModel):
    doi: Series[str] = pa.Field(nullable=True, alias=NAME_COL_DOI)
    title: Series[str] = pa.Field(nullable=True, alias=NAME_COL_TITLE)
    date: Series[datetime] = pa.Field(nullable=True, alias=NAME_COL_DATE)
    abstract: Series[str] = pa.Field(nullable=True, alias=NAME_COL_ABSTRACT)
    type: Optional[Series[Categorical]] = pa.Field(
        nullable=True, alias=NAME_COL_TYPE, categories=["article", "book"]
    )
    is_reference: Series[bool] = pa.Field(nullable=True, alias=NAME_COL_IS_REFERENCE)
    is_found: Series[bool] = pa.Field(nullable=True, alias=NAME_COL_IS_FOUND)
    year: Series[Int64] = pa.Field(nullable=True, alias=NAME_COL_YEAR)

    # Optional fields
    id_scopus: Optional[Series[str]] = pa.Field(nullable=True, alias=NAME_COL_ID_SCOPUS)
    id_openalex: Optional[Series[str]] = pa.Field(
        nullable=True, alias=NAME_COL_ID_OPENalex
    )
    id_lens: Optional[Series[str]] = pa.Field(nullable=True, alias=NAME_COL_ID_LENS)
    id_dimensions: Optional[Series[str]] = pa.Field(
        nullable=True, alias=NAME_COL_ID_DIMENSIONS
    )
    in_scopus: Optional[Series[bool]] = pa.Field(
        nullable=True, alias=NAME_COL_IN_SCOPUS
    )
    in_openalex: Optional[Series[bool]] = pa.Field(
        nullable=True, alias=NAME_COL_IN_OPENalex
    )
    in_lens: Optional[Series[bool]] = pa.Field(nullable=True, alias=NAME_COL_IN_LENS)
    in_dimensions: Optional[Series[bool]] = pa.Field(
        nullable=True, alias=NAME_COL_IN_DIMENSIONS
    )

    @pa.parser("type")  #
    def _parse_type(cls, series: Series) -> Series:
        result = series.copy()
        non_null_mask = series.notna()

        if non_null_mask.any():
            processed = series[non_null_mask].astype(str).str.lower().str.strip()
            processed = processed.map(TYPE_MAPPING).fillna("other")
            result.loc[non_null_mask] = processed

        return result


SchemaPapers(df_papers).dtypes
```
