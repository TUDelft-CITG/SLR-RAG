```python

def group_by_col_ref(
    df_input: pd.DataFrame, col_name: str, min_count_to_keep: int = 1
):
    df_gb_top = (
        df_input.groupby([col_name])
        .size()
        .sort_values(ascending=False)
        .rename("count")
        .reset_index()
    )

    df_gb_top = df_gb_top[(df_gb_top["count"] >= min_count_to_keep)]

    df_ret = (
        df_input.groupby([col_name, "is_reference"])
        .size()
        .sort_values(ascending=False)
        .rename("count")
        .reset_index()
    )
    df_ret = df_ret[df_ret[col_name].isin(df_gb_top[col_name].unique())]
    return df_ret
group_by_col_ref(df_test, "field_single")
```

```python
from typing import List, Set

def filter_on_list(
    df_input: pd.DataFrame, list_input: List[str] | Set[str], col_name: str, blacklist=True
):
    if blacklist:
        return df_input[df_input[col_name].isna() | (~df_input[col_name].isin(list_input))]
    else:
        return df_input[df_input[col_name].isna() | (df_input[col_name].isin(list_input))]
filter_on_list(df_test, {"Medicine", "Biology"}, "field_single",blacklist=True)
```

```python

def explode_and_filter_on_list(
    df_input: pd.DataFrame,
    col_name: str,
    black_list_input: Set[str] | List[str],
    seaparator=";",
    black_list=True
):
    """
    Filter all papers that have at least one keywords inside a blacklist
    """
    df_input = df_input.reset_index(drop=True)
    serie_is_topic_in_list = (
            df_input[col_name]
            .str
            .split(pat=seaparator)
            .explode()
            .str
            .strip()
            .isin(black_list_input)
            .groupby(level=0)
            .any()
        )
    if black_list:
        return df_input[(~serie_is_topic_in_list) | df_input[col_name].isna()]
    else:
        return df_input[serie_is_topic_in_list | df_input[col_name].isna()]
explode_and_filter_on_list(df_test,"fields_multiple",black_list_input={"AI", "Artificial Intelligence"},seaparator=",")
```

```python
def explode_and_groupby_topic(
    df_input: pd.DataFrame,
    name_topic_col: str,
    min_count_to_keep=1,
    separator=";",
):
    df_input = (
        df_input[df_input[name_topic_col].notna()]
        .reset_index(drop=True)
        .copy()
    )
    df_input[name_topic_col] = df_input[name_topic_col].str.split(pat=separator)
    df_input = df_input.explode(name_topic_col)
    df_input[name_topic_col] = df_input[name_topic_col].str.strip()
    return group_by_col_ref(df_input, name_topic_col, min_count_to_keep)
explode_and_groupby_topic(df_test,name_topic_col="fields_multiple",separator=",")
```

```python
import plotly.express as px
def plot_histogram(df_input: pd.DataFrame, col_name:str, min_count_to_keep = 1,seperator:str=None, height=500):
    if seperator is None:
        gb = group_by_col_ref(df_input,col_name,min_count_to_keep)
    else:
        gb = explode_and_groupby_topic(df_input,col_name,min_count_to_keep,seperator)
    return px.bar(
    gb,
    y=col_name,
    color="is_reference",
    x="count",
    height=height)
plot_histogram(df_test,"field_single")
```

```python
from pandera.typing import DataFrame
from tqdm import tqdm
from slrrag.settings import (
    NAME_COL_TITLE,
    NAME_COL_ABSTRACT,
    NAME_COL_IS_REFERENCE,
    NAME_COL_PRIMARY_TOPIC_OPENALEX,
)


def write_paper_into_markdown(
    df: DataFrame[SchemaPapers],
    nrows=50,
    output_dir=OUTPUT_PATH / "list_papers_to_analyse",
):
    """
    Write one markdown file per paper (first nrows) in the output directory.
    Each file contains the header and the paper's Title and Abstract.

    Args:
        df: DataFrame containing papers with 'Title' and 'Abstract' columns
        nrows: Number of rows to process (max files to create)
        output_dir: Directory to save the markdown files
    """

    # Ensure output directory exists
    Path(output_dir).mkdir(parents=True, exist_ok=True)

    # Take first nrows
    df_subset = df.head(nrows)

    for idx, row in tqdm(df_subset.iterrows(), total=len(df_subset)):
        title = row.get(NAME_COL_TITLE, f"untitled_{idx}")
        abstract = row.get(NAME_COL_ABSTRACT, "No Abstract Available")
        is_reference = row.get(NAME_COL_IS_REFERENCE, False)
        # Create a safe filename (replace problematic characters)

        filename = f"{idx}_paper.md"
        filepath = os.path.join(output_dir, filename)

        # Write the markdown file
        with open(filepath, "w", encoding="utf-8") as f:
            f.write("---\n")
            f.write("keywords_query:\n")
            f.write(f"is_reference: {'true' if is_reference else 'false'}\n")
            f.write("to_keep: false\n")
            f.write(f"topic_openalex: {row.get(NAME_COL_PRIMARY_TOPIC_OPENALEX, '')}\n")
            f.write("---\n\n")
            f.write(f"### {title}\n\n")
            f.write(f"{abstract}\n")


write_paper_into_markdown(df_recent.head(50))
```
