```python
from slrrag.toolbox import query_treemap

main_query = """
(
    (
        ("offshore") 
        AND 
        ("infrastructure" OR "construction" OR "installation" OR "windfarm" OR "wind farm" OR  "windmill" OR 
        "wind turbine" OR "wind park" OR "wind energy" OR "platform")
    )
    OR 
    ("flotel" OR "Aquaculture farm" OR "Artificial Island" OR "Synthetic Island")
)
AND
(
	( "collision" OR "allision" OR "near miss" OR "near-miss" OR "risk" OR "safety" OR "safe") AND 
    ( "navigate" OR "navigation" OR "navigational" OR "maritime" OR "shipping" OR "traffic")
	AND 
	("quantitative" OR "quantification" OR "probability" OR "probabilistic" OR "likelihood" 
        OR "frequency" OR "assess" OR "assessment" OR "predict" OR "predicting"
        OR "prediction" OR "model" OR "modeling" OR "modelling" OR "simulate"
        OR "simulation" OR "estimate" OR "estimation" OR "detect" OR "detection")
)
"""
tree = query_treemap.parse_query(main_query)
agg = query_treemap.aggregate_by_node(
    df=df_04,
    tree=tree,
    columns=["title", "abstract"],
    binary_columns=["is_reference"],
    stemming=False,
).sort_values("count", ascending=False)
fig = query_treemap.treemap(agg, width=1250, height=750)
fig
```

```python
def remove_brackets(df_input: pd.DataFrame, input_columns: list[str]):
    df_input_cp = df_input.copy()
    for col in input_columns:
        df_input_cp[col] = df_input_cp[col].map(
            lambda x: (
                [y.removesuffix("]]").removeprefix("[[") for y in x]
                if isinstance(x, list)
                else None
            )
        )
    return df_input_cp


df_papers_obsidian = pd.DataFrame.from_dict(vault.front_matter_index, orient="index")
df_papers_obsidian = remove_brackets(
    df_papers_obsidian, ["country", "authors", "source", "keywords"]
)
df_papers_obsidian
```

```python
main_colection_zotero_name = "SLR-RAG-LR-papers"

df_papers_obsidian_keep = df_papers_obsidian[
    df_papers_obsidian["collections"]
    .astype(str)
    .str.contains(main_colection_zotero_name + "-keep")
]

df_papers_obsidian_keep
```

```python
import country_converter as coco


def standardize_country_iso3(
    df: pd.DataFrame, column_name: str, standard="ISO3"
) -> pd.DataFrame:
    """Standardize a country-name column to ISO3 codes (e.g. 'Netherlands' -> 'NLD').

    Unmatched values become NaN. The original column is kept as
    f"{column_name}_original".
    """
    cc = coco.CountryConverter()
    df = df.copy()
    df[column_name + "_" + standard] = cc.convert(
        df[column_name].astype(str).str.strip(),
        to=standard,
        not_found=None,  # NaN instead of raising on unknown names
    )
    return df


df_papers_obsidian_keep_accessed_countries = standardize_country_iso3(
    df_papers_obsidian_keep_accessed_countries, "country"
)
df_papers_obsidian_keep_accessed_countries
```

```python
df_papers_countries_count = pd.DataFrame(
    df_papers_obsidian_keep_accessed_countries.groupby("country_ISO3")
    .size()
    .sort_values(ascending=False)
    .rename("number_papers")
)
df_papers_countries_count
```

```python
from slrrag.settings import PROJECT_ROOT
from pathlib import Path

path_offfshore_wind_power_mw_gem_csv = Path(
    PROJECT_ROOT
    / "data"
    / "gem_wind_power_tracker"
    / "Global Wind Power Tracker Offshore Wind Summary Data - Feb 2026 - Offshore Wind Farms by Country_Area (MW).csv"
)
path_offfshore_wind_power_mw_gem_csv
path_offfshore_wind_power_mw_gem_csv.exists()
```

```python
import pandas as pd

df_offshore_wind_power_mw = pd.read_csv(
    path_offfshore_wind_power_mw_gem_csv, skiprows=4, thousands=","
)
columns_df_offshore_wind_power_mw = [
    "country_area",
    "operating",
    "construction",
    "pre_construction",
    "announced",
    "prospective",
    "cancelled",
    "shelved",
    "retired",
    "mothballed",
]

df_offshore_wind_power_mw.columns = columns_df_offshore_wind_power_mw
df_offshore_wind_power_mw = df_offshore_wind_power_mw.sort_values(
    "prospective", ascending=False
)
# Remove the global total line
df_offshore_wind_power_mw = df_offshore_wind_power_mw[1:]
df_offshore_wind_power_mw
```

```python
import pandas as pd


df_offshore_wind_power_mw = standardize_country_iso3(
    df_offshore_wind_power_mw, "country_area"
)
df_offshore_wind_power_mw
```

```python
df_offshore_wind_power_mw.merge(
    df_papers_countries_count,
    left_on="country_area_ISO3",
    right_on="country_ISO3",
    how="outer",
).fillna({"number_papers": 0}).astype({"number_papers": int})
```

```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots

status_columns = [
    "operating",
    "construction",
    "pre_construction",
    "announced",
    "prospective",
]
df_plot = (
    df_offshore_wind_power_mw.merge(
        df_papers_countries_count,
        left_on="country_area_ISO3",
        right_on="country_ISO3",
        how="outer",
    )
    .fillna({"number_papers": 0})
    .astype({"number_papers": int})
    .sort_values("number_papers", ascending=False)
    .query(
        "(operating > 0) | (construction > 0) | (pre_construction > 0)| (number_papers > 0)"
    )
    .query("number_papers > 0")
)

fig = make_subplots(specs=[[{"secondary_y": True}]])

for status in status_columns:
    fig.add_trace(
        go.Bar(
            x=df_plot["country_area"],
            y=df_plot[status],
            name=status.replace("_", " ").title(),
        ),
        secondary_y=False,
    )

fig.add_trace(
    go.Scatter(
        x=df_plot["country_area"],
        y=df_plot["number_papers"],  # <-- your paper-count column
        name="Number of papers",
        mode="lines+markers",
        line=dict(color="black", width=2),
        marker=dict(size=8),
    ),
    secondary_y=True,
)

fig.update_layout(
    barmode="group",
    title="Offshore wind capacity by status and number of papers per country",
    xaxis_tickangle=-45,
)
fig.update_yaxes(title_text="Capacity (MW)", secondary_y=False)
fig.update_yaxes(title_text="Number of papers", secondary_y=True, showgrid=False)
fig.show()
```
