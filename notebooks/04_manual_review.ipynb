```python
%load_ext autoreload
%autoreload 2
import pandas as pd

from slrrag.main import merge_papers
from slrrag.settings import PATH_CSV_03_AGENTS_ANWSERS, PATH_CSV_04_MANUAL_FILTER
from slrrag.toolbox.files import slugify

df_04_manual_filter = pd.read_csv(PATH_CSV_04_MANUAL_FILTER)
df_04_manual_filter
```

```python
df_03_papers = pd.read_csv(PATH_CSV_03_AGENTS_ANWSERS)
df_03_papers["title_normal"] = df_03_papers["title"]
df_03_papers["title_slug"] = df_03_papers["title"].apply(slugify)
df_03_papers
```

```python
df_03_papers.groupby("is_reference", dropna=False).size()
```

```python
df_03_papers = merge_papers(
    df_03_papers, df_04_manual_filter[["title", "doi", "tags"]], "manual"
)
df_03_papers = df_03_papers[~df_03_papers["tags"].isna()]
df_03_papers
```

```python
df_03_papers[df_03_papers["is_reference"]]
```

```python
df_03_papers["user_keep"] = df_03_papers["tags"].isin(["#keep"])
df_03_papers
```

```python
from slrrag.input import FIRST_ROUND_QUESTIONS, OLLAMA_MODELS

type_list_col_answer = [
    q.get("name") + "_answer_" + x for x in OLLAMA_MODELS for q in FIRST_ROUND_QUESTIONS
]
type_list_col_justification = [
    q.get("name") + "_justification_" + x
    for x in OLLAMA_MODELS
    for q in FIRST_ROUND_QUESTIONS
]
col_quest = df_03_papers.columns[
    df_03_papers.columns.str.startswith(
        tuple(type_list_col_answer + type_list_col_justification)
    )
]
df_03_papers[
    ~(df_03_papers["user_keep"] == df_03_papers[col_quest].any(axis=1))
    & (df_03_papers["tags"] != "#not-sure")
][["title"] + list(col_quest) + ["user_keep", "tags"]]
```

```python
df_03_papers[list(col_quest)].any(axis=1).describe()
```

### Comparison with selection

```python
from slrrag.settings import PATH_CSV_06

df_06_picky_selection = pd.read_csv(PATH_CSV_06)
df_06_picky_selection
```

```python
df_06_picky_selection["title_slug"] = df_06_picky_selection["title"].apply(slugify)
df_merge = pd.merge(
    df_06_picky_selection[
        [
            "title_slug",
            "nra_lr_classification",
            "nra_lr_perspective",
            "nra_lr_refusal_resons",
        ]
    ],
    df_03_papers,
    how="inner",
    on="title_slug",
)
df_merge
```

```python
df_merge["is_picky_kept"] = ~df_merge["nra_lr_classification"].isna()
df_merge
```

#### With Dash

```python
type_list_col_answer = [
    q.get("name") + "_answer_" + x for x in OLLAMA_MODELS for q in FIRST_ROUND_QUESTIONS
]
type_list_col_justification = [
    q.get("name") + "_justification_" + x
    for x in OLLAMA_MODELS
    for q in FIRST_ROUND_QUESTIONS
]
type_list_col_answer, type_list_col_justification
```

```python
df_dash_plot = df_merge[
    ["title", "is_picky_kept"] + type_list_col_answer + type_list_col_justification
]

df_dash_plot
```

```python
from slrrag.settings import PATH_INPUT

df_dash_plot["consensus_score"] = (
    df_dash_plot[type_list_col_answer].apply(
        lambda x: x == df_dash_plot["is_picky_kept"]
    )
).sum(axis=1) / 3
df_dash_plot = df_dash_plot[
    ["title", "consensus_score", "is_picky_kept"]
    + type_list_col_answer
    + type_list_col_justification
]
df_dash_plot = df_dash_plot.sort_values("consensus_score", ascending=True)
df_dash_plot.to_csv(PATH_INPUT / "result_agents_to_plot.csv")
df_dash_plot.dtypes
```

```python
from dash import Dash

from slrrag.toolbox.dash_app import create_app_layout

binary_columns = type_list_col_answer + ["is_picky_kept"]
app = Dash(__name__)

app.layout = (
    create_app_layout(df_dash_plot.dropna())
    .add_data_table_tab()
    .add_performance_metrics_tab(
        model_columns=type_list_col_answer, y_true_col="is_picky_kept"
    )
    .add_roc_curve_tab(model_columns=type_list_col_answer)
    .add_upset_plot_tab(binary_columns=binary_columns, mode="distinct")
    .build()
)

app.run(jupyter_mode="external", debug=True)
```
