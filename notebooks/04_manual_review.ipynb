#### With Dash

```python
type_list_col_answer = [
    q.get("name") + "_answer_" + x for x in OLLAMA_MODELS for q in FIRST_ROUND_QUESTIONS
]
type_list_col_justification = [
    q.get("name") + "_justification_" + x
    for x in OLLAMA_MODELS
    for q in FIRST_ROUND_QUESTIONS
]
type_list_col_answer, type_list_col_justification
```

```python
df_dash_plot = df_merge[
    ["title", "is_picky_kept"] + type_list_col_answer + type_list_col_justification
]

df_dash_plot
```

```python
from slrrag.settings import PATH_INPUT

df_dash_plot["consensus_score"] = (
    df_dash_plot[type_list_col_answer].apply(
        lambda x: x == df_dash_plot["is_picky_kept"]
    )
).sum(axis=1) / 3
df_dash_plot = df_dash_plot[
    ["title", "consensus_score", "is_picky_kept"]
    + type_list_col_answer
    + type_list_col_justification
]
df_dash_plot = df_dash_plot.sort_values("consensus_score", ascending=True)
df_dash_plot.to_csv(PATH_INPUT / "result_agents_to_plot.csv")
df_dash_plot.dtypes
```

```python
import pandas as pd
from dash import Dash, dash_table, html

app = Dash(__name__)

# Sort by consensus score (descending - highest first)
# Binary columns to color (green for 1, red for 0)
binary_columns = type_list_col_answer + ["is_picky_kept"]

# Build conditional styling
style_data_conditional = []

# Binary columns: green for 1, red for 0
for col in binary_columns:
    style_data_conditional.append(
        {
            "if": {"filter_query": f"{{{col}}} contains true", "column_id": col},
            "backgroundColor": "#90EE90",
            "color": "black",
        }
    )
    style_data_conditional.append(
        {
            "if": {"filter_query": f"{{{col}}} contains false", "column_id": col},
            "backgroundColor": "#FFCCCB",
            "color": "black",
        }
    )

# consensus_score column: gradient from red (0) to green (1)
# Using multiple conditions to simulate a gradient
style_data_conditional.append(
    {
        "if": {
            "filter_query": "{consensus_score} < 0.2",
            "column_id": "consensus_score",
        },
        "backgroundColor": "#FF0000",
        "color": "white",
    }
)
style_data_conditional.append(
    {
        "if": {
            "filter_query": "{consensus_score} >= 0.2 && {consensus_score} < 0.4",
            "column_id": "consensus_score",
        },
        "backgroundColor": "#FF4500",
        "color": "white",
    }
)
style_data_conditional.append(
    {
        "if": {
            "filter_query": "{consensus_score} >= 0.4 && {consensus_score} < 0.6",
            "column_id": "consensus_score",
        },
        "backgroundColor": "#FFA500",
        "color": "black",
    }
)
style_data_conditional.append(
    {
        "if": {
            "filter_query": "{consensus_score} >= 0.6 && {consensus_score} < 0.8",
            "column_id": "consensus_score",
        },
        "backgroundColor": "#9ACD32",
        "color": "black",
    }
)
style_data_conditional.append(
    {
        "if": {
            "filter_query": "{consensus_score} >= 0.8",
            "column_id": "consensus_score",
        },
        "backgroundColor": "#00FF00",
        "color": "black",
    }
)

app.layout = html.Div(
    children=[
        html.H1(children="SLR-RAG Data Table"),
        html.Div(
            children="""
        Dash table displaying the review data.
    """
        ),
        dash_table.DataTable(
            id="data-table",
            columns=[{"name": col, "id": col} for col in df_dash_plot.columns],
            data=df_dash_plot.to_dict("records"),
            style_table={"overflowX": "auto"},
            style_cell={
                "textAlign": "left",
                "padding": "10px",
                "whiteSpace": "normal",
                "height": "auto",
            },
            style_header={
                "backgroundColor": "rgb(230, 230, 230)",
                "fontWeight": "bold",
            },
            style_data_conditional=style_data_conditional,
            page_size=10,
        ),
    ]
)

app.run(jupyter_mode="external", debug=True)
```
