### This notebook is for comparing manual review made on obsidian, and agents analysis

```python
%load_ext autoreload
%autoreload 2
import pandas as pd

from slrrag.main import merge_papers
from slrrag.settings import PATH_CSV_03_AGENTS_ANWSERS, PATH_INPUT
from slrrag.toolbox.files import slugify
ollama_models = [
"granite4.1:30b",
"ministral-3:14b",
"llama3.1:8b",
]

local_path_csv_04_manual_filter = PATH_INPUT / "04_papers_manual_filter_query_23_v2.csv"
local_path_csv_03_agents_anwsers = PATH_CSV_03_AGENTS_ANWSERS
df_04_manual_filter = pd.read_csv(local_path_csv_04_manual_filter).rename(columns={"title": "title_slug"})
df_04_manual_filter
```

```python
# May be slighlty more than 04, because of duplicates.
df_03_papers = pd.read_csv(local_path_csv_03_agents_anwsers)
df_03_papers["title_slug"] = df_03_papers["title"].apply(slugify)
df_03_papers
```

```python
df_04_manual_filter = df_04_manual_filter[df_04_manual_filter["title_slug"].isin(df_03_papers["title_slug"])]
df_04_manual_filter
```

```python
df_04_merged = merge_papers(
    df_03_papers,
    df_04_manual_filter[["title_slug", "doi", "tags"]],
    "manual",
    name_col_title="title_slug",
    need_to_slugify=False,
)
df_04_merged = df_04_merged[~df_04_merged["tags"].isna()]
df_04_merged
```

```python
answer_human_name_col = "answer_human"
df_04_merged[answer_human_name_col] = df_04_merged["tags"].isin(["#keep"])
df_04_merged
```

```python
type_list_col_answer = ["answer_" + x for x in ollama_models
]
type_list_col_justification = ["explanation_" + x
    for x in ollama_models
]
df_04_merged[type_list_col_answer].describe()
```

```python
from slrrag.settings import PATH_INPUT
human_consensus_score_column_name = "human_consensus_score"
df_04_merged[human_consensus_score_column_name] = (
    df_04_merged[type_list_col_answer].apply(
        lambda x: x == df_04_merged[answer_human_name_col]
    )
).sum(axis=1) / len(ollama_models)
df_04_merged[human_consensus_score_column_name].describe()
```

## Dash plot

```python
from dash import Dash

from slrrag.toolbox.dash_app import create_app_layout

df_dash_plot = df_04_merged[
    ["title", human_consensus_score_column_name, answer_human_name_col]
    + type_list_col_answer
    + type_list_col_justification
]
df_dash_plot = df_dash_plot.sort_values(human_consensus_score_column_name, ascending=True)

binary_columns = type_list_col_answer + [answer_human_name_col]
app = Dash(__name__)

app.layout = (
    create_app_layout(df_dash_plot.dropna())
    .add_data_table_tab()
    .add_performance_metrics_tab(
        model_columns=type_list_col_answer, y_true_col=answer_human_name_col
    )
    .add_roc_curve_tab(model_columns=type_list_col_answer,y_true_col=answer_human_name_col)
    .add_upset_plot_tab(binary_columns=binary_columns, mode="distinct")
    .build()
)

app.run(jupyter_mode="external", debug=True)
```
