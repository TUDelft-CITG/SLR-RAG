```python
%load_ext autoreload
%autoreload 2
import pandas as pd
from slrrag.settings import PATH_CSV_04_MANUAL_FILTER, PATH_CSV_03_AGENTS_ANWSERS
from slrrag.main import merge_papers
from slrrag.toolbox.files import slugify
from slrrag.upset_plotter import plot_upset

df_manual_filter = pd.read_csv(PATH_CSV_04_MANUAL_FILTER)
df_manual_filter
```

```python
df_papers = pd.read_csv(PATH_CSV_03_AGENTS_ANWSERS)
df_papers["title_normal"] = df_papers["title"]
df_papers["title"] = df_papers["title"].apply(slugify)
df_papers
```

```python
df_papers = merge_papers(
    df_papers, df_manual_filter[["title", "doi", "tags"]], "manual"
)
df_papers = df_papers[~df_papers["tags"].isna()]
df_papers
```

```python
df_papers[df_papers["is_reference"]]
```

```python
plot_upset(df_papers[df_papers["tags"] == "#keep"])
```

```python
df_papers["user_keep"] = df_papers["tags"].isin(["#keep"])
df_papers
```

```python
from slrrag.input import FIRST_ROUND_QUESTIONS, OLLAMA_MODELS


type_list_col_answer = [
    q.get("name") + "_answer_" + x for x in OLLAMA_MODELS for q in FIRST_ROUND_QUESTIONS
]
type_list_col_justification = [
    q.get("name") + "_justification_" + x
    for x in OLLAMA_MODELS
    for q in FIRST_ROUND_QUESTIONS
]
col_quest = df_papers.columns[
    df_papers.columns.str.startswith(
        tuple(type_list_col_answer + type_list_col_justification)
    )
]
df_papers[
    ~(df_papers["user_keep"] == df_papers[col_quest].any(axis=1))
    & (df_papers["tags"] != "#not-sure")
][["title"] + list(col_quest) + ["user_keep", "tags"]]
```
